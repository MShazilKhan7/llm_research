Run all models × strategies × tasks and save results to `results/`.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))

# ── SET YOUR API KEYS HERE ──────────────────────────────────
os.environ['OPENAI_API_KEY']  = 'sk-...'   # OpenAI / GPT
os.environ['GOOGLE_API_KEY']  = '...'      # Google Gemini
os.environ['GROQ_API_KEY']    = 'gsk_...'  # Groq (Llama)
# ────────────────────────────────────────────────────────────

MOCK = False   # Set to True to run without real API calls
DELAY = 1.0    # Seconds between API calls (respect rate limits)

DATASET_PATH = '../dataset/ground_truth.csv'
RESULTS_DIR  = '../results'

print('Config ready.')
```

```python
from scripts.run_inference import load_dataset, run_combination
from scripts.llm_clients import SUPPORTED_MODELS
from prompts.prompt_templates import TASKS, PROMPT_STRATEGIES

issues = load_dataset(DATASET_PATH)
print(f'Loaded {len(issues)} issues')
```

## Run Inference (all combinations)

3 models × 2 tasks × 4 strategies = **24 runs**

```python
completed = []

for model in SUPPORTED_MODELS:
    for task in TASKS:
        for strategy in PROMPT_STRATEGIES:
            out = run_combination(
                issues=issues,
                model_name=model,
                task=task,
                strategy=strategy,
                mock=MOCK,
                delay=DELAY,
                output_dir=RESULTS_DIR,
            )
            completed.append(out)

print(f'\n✓ Done — {len(completed)} result files saved to {RESULTS_DIR}/')
```
