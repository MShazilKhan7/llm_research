## Inspect one ensemble file

```python
import pandas as pd

df = pd.read_csv(f'{RESULTS_DIR}/ensemble_ambiguity_zero_shot.csv')
df.head(10)
```

```python
# Agreement rate — how often do all 3 models agree?
pred_cols = [c for c in df.columns if c.startswith('pred_')]
df['all_agree'] = df[pred_cols].nunique(axis=1) == 1
print(f'Full agreement rate: {df["all_agree"].mean():.2%}')

import matplotlib.pyplot as plt
df['n_yes_votes'] = df[pred_cols].sum(axis=1)
df['n_yes_votes'].value_counts().sort_index().plot(
    kind='bar', color='steelblue', alpha=0.8, figsize=(6, 3)
)
plt.title('Distribution of Yes Votes per Issue (Ambiguity / Zero-shot)')
plt.xlabel('Number of models predicting Yes')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
