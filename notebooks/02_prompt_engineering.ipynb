# Notebook 2 — Prompt Engineering Preview

This notebook lets you inspect all four prompting strategies for both tasks
without needing any API key.


```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))

from prompts.prompt_templates import PROMPT_REGISTRY, PROMPT_STRATEGIES, TASKS
```

```python
# Sample issue to preview
SAMPLE_TITLE = 'Fix login bug'
SAMPLE_DESCRIPTION = 'Users cannot log in sometimes. Please fix it.'

print(f'Issue Title      : {SAMPLE_TITLE}')
print(f'Issue Description: {SAMPLE_DESCRIPTION}')
```

## Ambiguity Prompts

```python
for strategy in PROMPT_STRATEGIES:
    prompt_fn = PROMPT_REGISTRY['ambiguity'][strategy]
    prompt = prompt_fn(SAMPLE_TITLE, SAMPLE_DESCRIPTION)
    print('=' * 60)
    print(f'Strategy: {strategy.upper()}')
    print('=' * 60)
    print(prompt)
    print()
```

## Incompleteness Prompts

```python
for strategy in PROMPT_STRATEGIES:
    prompt_fn = PROMPT_REGISTRY['incompleteness'][strategy]
    prompt = prompt_fn(SAMPLE_TITLE, SAMPLE_DESCRIPTION)
    print('=' * 60)
    print(f'Strategy: {strategy.upper()}')
    print('=' * 60)
    print(prompt)
    print()
```

## Token length estimate

```python
import pandas as pd

rows = []
for task in TASKS:
    for strategy in PROMPT_STRATEGIES:
        prompt = PROMPT_REGISTRY[task][strategy](SAMPLE_TITLE, SAMPLE_DESCRIPTION)
        # Rough token estimate: words * 1.3
        tokens = int(len(prompt.split()) * 1.3)
        rows.append({'task': task, 'strategy': strategy,
                     'chars': len(prompt), 'est_tokens': tokens})

pd.DataFrame(rows)
```
