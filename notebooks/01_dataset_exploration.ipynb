## 3. Label Distribution

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

for ax, col, title, colors in zip(
    axes,
    ['ambiguous', 'incomplete'],
    ['Ambiguity Labels', 'Incompleteness Labels'],
    [['#55A868','#C44E52'], ['#4C72B0','#DD8452']]
):
    counts = df[col].value_counts().sort_index()
    ax.bar(['Not Defective (0)', 'Defective (1)'], counts.values, color=colors, alpha=0.85)
    for i, v in enumerate(counts.values):
        ax.text(i, v + 1, str(v), ha='center', fontweight='bold')
    ax.set_title(title, fontsize=13, fontweight='bold')
    ax.set_ylabel('Count')
    ax.set_ylim(0, max(counts.values) * 1.15)
    ax.yaxis.grid(True, alpha=0.4)
    ax.set_axisbelow(True)

plt.suptitle('Ground Truth Label Distribution (200 Issues)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('../figures/label_distribution.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Cross-tabulation: how many issues are both ambiguous AND incomplete?
cross = pd.crosstab(df['ambiguous'], df['incomplete'],
                    rownames=['Ambiguous'], colnames=['Incomplete'])
print('Cross-tabulation:')
cross
```
